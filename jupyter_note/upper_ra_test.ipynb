```python
import sys
sys.path.append('../src/')

import my_functions
from my_functions import binning, cutting
from astropy.table import Table
from pixell import enmap
import thumbstack
from thumbstack import ThumbStack
```

```python
this_bin = binning('FujiBGSPhysProp_v1.1.fits', 4, 'RA') # splits data into four bins based on RA

cutting('FujiBGSPhysProp_v1.1.fits', {'RA': [this_bin['RA_min'][3], this_bin['RA_max'][3]]}, save=True, name='upper_ra') # creates new set for the upper ra
```

```python
this_data = Table.read('upper_ra.fits', format='fits')
this_map = enmap.read_map('re_runs/150_night_map.fits')
this_mask = enmap.read_map('re_runs/150_night_60_mask.fits')

ts = ThumbStack(this_data, 
                this_map, 
                this_mask, 
                name='upper_ra1',
                nameLong=None, 
                save=False, 
                nProc=32,
                filterTypes='ringring2',
                #doMBins=False, 
                doBootstrap=True,
                # doStackedMap=True,
                #doVShuffle=False, 
                #cmbNu=cmap.nu, 
                #cmbUnitLatex=cmap.unitLatex,
                pathOut='')
```
